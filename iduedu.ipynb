```python
import networkx as nx
import re
from collections import defaultdict

from collections import defaultdict
import re

def filter_routes_one_direction(graph):
    """
    Фильтрует маршруты, оставляя только одно направление для каждого маршрута,
    игнорируя порядок начальной и конечной точки.
    """
    # Собираем все уникальные маршруты из узлов типа platform
    all_routes = set()
    
    for node, data in graph.nodes(data=True):
        if data.get('type') == 'platform':
            route = data.get('route', [])
            if isinstance(route, str):
                all_routes.add(route)
            elif isinstance(route, list):
                all_routes.update(route)
    
    # Группируем маршруты по базовому имени (без учета направления)
    route_groups = defaultdict(list)
    
    for route in all_routes:
        # Извлекаем базовое имя маршрута (до символов направления)
        # Ищем паттерн: "Тип № номер: откуда => куда" или "Тип № номер: откуда — куда"
        match = re.match(r'^([^:]+):\s*(.+)\s*(?:=>|—)\s*(.+)$', route)
        if match:
            transport_type = match.group(1).strip()  # "Автобус № 220"
            start_point = match.group(2).strip()
            end_point = match.group(3).strip()
            
            # Создаем базовый ключ, сортируя точки по алфавиту
            sorted_points = sorted([start_point, end_point])
            base_key = f"{transport_type}:{sorted_points[0]}|{sorted_points[1]}"
            route_groups[base_key].append(route)
    
    # Выбираем по одному маршруту из каждой группы
    selected_routes = set()
    for group in route_groups.values():
        if len(group) > 1:
            # Если есть несколько направлений, выбираем первый
            selected_routes.add(group[0])
        else:
            selected_routes.add(group[0])
    
    return selected_routes

def filter_nodes_by_routes(graph, selected_routes):
    """
    Фильтрует узлы, оставляя только те, которые относятся к выбранным маршрутам
    """
    nodes_to_keep = set()
    
    for node, data in graph.nodes(data=True):
        node_type = data.get('type')
        
        if node_type == 'platform':
            route = data.get('route', [])
            routes_list = [route] if isinstance(route, str) else route
            
            # Проверяем, есть ли пересечение с выбранными маршрутами
            if any(r in selected_routes for r in routes_list):
                nodes_to_keep.add(node)
        else:
            # Оставляем все остальные узлы (остановки)
            nodes_to_keep.add(node)
    
    # Создаем подграф с отфильтрованными узлами
    filtered_graph = graph.subgraph(nodes_to_keep).copy()
    return filtered_graph

def redirect_edges_to_platforms(graph, selected_routes):
    """
    Перенаправляет ребра с остановок на соответствующие платформы
    """
    # Создаем мапинг: остановка -> список платформ для каждого маршрута
    stop_to_platforms = defaultdict(lambda: defaultdict(list))
    
    # Собираем информацию о boarding ребрах
    for u, v, data in graph.edges(data=True):
        if data.get('type') == 'boarding':
            route = data.get('route', '')
            if route in selected_routes:
                # u - остановка, v - платформа
                stop_to_platforms[u][route].append(v)
    
    # Создаем новый граф
    new_graph = nx.MultiDiGraph()
    
    # Копируем все узлы
    for node, data in graph.nodes(data=True):
        new_graph.add_node(node, **data)
    
    # Обрабатываем ребра
    for u, v, data in graph.edges(data=True):
        edge_type = data.get('type', '')
        route = data.get('route', '')
        
        if edge_type == 'boarding':
            # Пропускаем boarding ребра - они больше не нужны
            continue
        elif edge_type in ['bus', 'trolleybus', 'metro', 'tram'] and route in selected_routes:
            # Перенаправляем ребра транспорта с остановок на платформы
            
            # Находим платформы для начальной и конечной остановки
            u_platforms = stop_to_platforms[u].get(route, [])
            v_platforms = stop_to_platforms[v].get(route, [])
            
            if u_platforms and v_platforms:
                # Соединяем все платформы начальной остановки со всеми платформами конечной
                for u_platform in u_platforms:
                    for v_platform in v_platforms:
                        new_graph.add_edge(u_platform, v_platform, **data)
            else:
                # Если нет соответствующих платформ, оставляем оригинальное ребро
                new_graph.add_edge(u, v, **data)
        else:
            # Копируем остальные ребра как есть
            if route in selected_routes or not route:  # Включаем ребра без маршрута
                new_graph.add_edge(u, v, **data)
    
    return new_graph

def process_transport_graph(graph):
    """
    Основная функция для обработки графа общественного транспорта
    """
    print("Начинаем обработку графа...")
    
    # Шаг 1: Выбираем маршруты только в одну сторону
    print("Шаг 1: Фильтрация маршрутов по направлению...")
    selected_routes = filter_routes_one_direction(graph)
    print(f"Выбрано маршрутов: {len(selected_routes)}")
    
    # Выводим несколько примеров выбранных маршрутов
    print("Примеры выбранных маршрутов:")
    for i, route in enumerate(list(selected_routes)[:5]):
        print(f"  {i+1}. {route}")
    
    # Шаг 2: Фильтруем узлы по выбранным маршрутам
    print("Шаг 2: Фильтрация узлов...")
    filtered_graph = filter_nodes_by_routes(graph, selected_routes)
    print(f"Узлов до фильтрации: {graph.number_of_nodes()}")
    print(f"Узлов после фильтрации: {filtered_graph.number_of_nodes()}")
    
    # Шаг 3: Перенаправляем ребра с остановок на платформы
    print("Шаг 3: Перенаправление ребер на платформы...")
    final_graph = redirect_edges_to_platforms(filtered_graph, selected_routes)
    print(f"Ребер до перенаправления: {filtered_graph.number_of_edges()}")
    print(f"Ребер после перенаправления: {final_graph.number_of_edges()}")
    
    return final_graph, selected_routes

# Пример использования:
# processed_graph, selected_routes = process_transport_graph(your_graph)

# Функция для анализа результата
def analyze_graph(graph):
    """
    Анализирует структуру обработанного графа
    """
    print("\n=== Анализ обработанного графа ===")
    
    # Подсчет узлов по типам
    node_types = defaultdict(int)
    for node, data in graph.nodes(data=True):
        node_type = data.get('type', 'unknown')
        node_types[node_type] += 1
    
    print("Количество узлов по типам:")
    for node_type, count in node_types.items():
        print(f"  {node_type}: {count}")
    
    # Подсчет ребер по типам
    edge_types = defaultdict(int)
    for u, v, data in graph.edges(data=True):
        edge_type = data.get('type', 'unknown')
        edge_types[edge_type] += 1
    
    print("Количество ребер по типам:")
    for edge_type, count in edge_types.items():
        print(f"  {edge_type}: {count}")
    
    # Проверяем связность
    if graph.is_directed():
        weakly_connected = nx.is_weakly_connected(graph)
        print(f"Слабо связный граф: {weakly_connected}")
        if not weakly_connected:
            components = list(nx.weakly_connected_components(graph))
            print(f"Количество слабо связных компонент: {len(components)}")
    else:
        connected = nx.is_connected(graph)
        print(f"Связный граф: {connected}")
        if not connected:
            components = list(nx.connected_components(graph))
            print(f"Количество связных компонент: {len(components)}")
```

```python
import folium
import geopandas as gpd
from IPython.display import display, HTML
import ast  # For parsing string tuples

# Assuming points_gdf and roads_gdf are already reprojected to EPSG:4326

def select_route(route_number):
    """
    Filter points_gdf and roads_gdf by a given route number.
    
    Parameters:
    route_number (str): The route number to filter by (e.g., '158', '220').
    
    Returns:
    tuple: (filtered_points_gdf, filtered_roads_gdf)
    """
    # Convert route_number to string to handle numeric inputs
    route_number = str(route_number)
    
    # Filter points_gdf: include rows where route_number is in the route column
    points_filtered = points_gdf[
        points_gdf['route'].apply(lambda x: route_number in str(x) if isinstance(x, (list, str)) else False)
    ]
    
    # Filter roads_gdf: include rows where route_number is in the route column
    roads_filtered = roads_gdf[
        roads_gdf['route'].apply(lambda x: route_number in str(x) if isinstance(x, (list, str)) else False)
    ]
    
    return points_filtered, roads_filtered

def calculate_point_degrees(points_filtered, roads_filtered):
    """
    Calculate the degree of each point based on connected edges.
    
    Parameters:
    points_filtered (GeoDataFrame): Filtered points GeoDataFrame.
    roads_filtered (GeoDataFrame): Filtered roads GeoDataFrame.
    
    Returns:
    dict: Dictionary mapping point index to degree.
    """
    degree_dict = {row['index']: 0 for idx, row in points_filtered.iterrows()}
    
    # Check if roads_filtered has a MultiIndex or uses 'index' column
    for idx, row in roads_filtered.iterrows():
        # Try to get start and end nodes from the index or 'index' column
        if isinstance(idx, tuple):
            start_node, end_node = idx
        elif 'index' in row and isinstance(row['index'], (str, tuple)):
            # Handle 'index' column: could be a string like "(2669, 2670.0)" or a tuple
            if isinstance(row['index'], str):
                try:
                    # Parse string representation of tuple, e.g., "(2669, 2670.0)"
                    start_node, end_node = ast.literal_eval(row['index'])
                except (ValueError, SyntaxError):
                    continue  # Skip if index cannot be parsed
            else:
                start_node, end_node = row['index']
        else:
            # Skip if no valid start/end nodes can be determined
            continue
        
        # Convert nodes to float to match points_gdf index type
        start_node = float(start_node) if isinstance(start_node, (int, float)) else start_node
        end_node = float(end_node) if isinstance(end_node, (int, float)) else end_node
        
        # Increment degree for start and end nodes
        if start_node in degree_dict:
            degree_dict[start_node] += 1
        if end_node in degree_dict:
            degree_dict[end_node] += 1
    
    return degree_dict

def visualize_route(points_filtered, roads_filtered, route_number):
    """
    Visualize filtered points and roads for a given route on a Folium map.
    Points are colored by degree, and a legend is added.
    
    Parameters:
    points_filtered (GeoDataFrame): Filtered points GeoDataFrame.
    roads_filtered (GeoDataFrame): Filtered roads GeoDataFrame.
    route_number (str): The route number for naming the map.
    
    Returns:
    folium.Map: The generated Folium map object.
    """
    points_filtered.to_crs(4326,inplace=True)
    roads_filtered.to_crs(4326,inplace=True)
    # Initialize a Folium map
    if not points_filtered.empty:
        center = [points_filtered.geometry.y.mean(), points_filtered.geometry.x.mean()]
    elif not roads_filtered.empty:
        center = [roads_filtered.geometry.centroid.y.mean(), roads_filtered.geometry.centroid.x.mean()]
    else:
        center = [0, 0]  # Default center if no data
        print(f"No data found for route {route_number}")
        return None
    
    m = folium.Map(location=center, zoom_start=12, tiles='OpenStreetMap')
    
    # Calculate point degrees
    degree_dict = calculate_point_degrees(points_filtered, roads_filtered)
    
    # Define color scheme based on degree
    def get_color(degree):
        if degree <= 1:
            return 'green'  # Low degree
        elif degree <= 3:
            return 'orange'  # Medium degree
        else:
            return 'red'  # High degree
    
    # Add points to the map as markers, colored by degree
    if not points_filtered.empty:
        point_group = folium.FeatureGroup(name=f"Points (Route {route_number})").add_to(m)
        for idx, row in points_filtered.iterrows():
            degree = degree_dict.get(row['index'], 0)
            color = get_color(degree)
            folium.CircleMarker(
                location=[row.geometry.y, row.geometry.x],  # y=latitude, x=longitude
                radius=5,
                color=color,
                fill=True,
                fill_color=color,
                fill_opacity=0.6,
                popup=f"Type: {row['type']}<br>Route: {row['route']}<br>Index: {row['index']}<br>Degree: {degree}"
            ).add_to(point_group)
    
    # Add roads to the map as lines
    if not roads_filtered.empty:
        road_group = folium.FeatureGroup(name=f"Roads (Route {route_number})").add_to(m)
        for idx, row in roads_filtered.iterrows():
            coords = [[point[1], point[0]] for point in row.geometry.coords]  # [lat, lon]
            folium.PolyLine(
                locations=coords,
                color='red',
                weight=3,
                opacity=0.8,
                popup=f"Type: {row['type']}<br>Route: {row['route']}<br>Length: {row['length_meter']:.2f} m<br>Time: {row['time_min']:.2f} min<br>Index: {row['index']}"
            ).add_to(road_group)
    
    # Add legend for point colors
    legend_html = """
    <div style="position: fixed; bottom: 50px; left: 50px; z-index:9999; font-size:14px;
                background-color:white; padding:10px; border:2px solid grey; border-radius:5px;">
        <b>Point Degree Legend</b><br>
        <i style="color:green;">●</i> Low Degree (≤1)<br>
        <i style="color:orange;">●</i> Medium Degree (2-3)<br>
        <i style="color:red;">●</i> High Degree (>3)
    </div>
    """
    m.get_root().html.add_child(folium.Element(legend_html))
    
    # Add layer control to toggle points and roads
    folium.LayerControl().add_to(m)
    
    
    return m

# Example usage: Select and visualize route '158'
route_number = '31'
points_filtered, roads_filtered = select_route(route_number)
visualize_route(points_filtered, roads_filtered, route_number)

# To visualize a different route, call select_route and visualize_route with the desired route number
# e.g., points_filtered, roads_filtered = select_route('220')
#       visualize_route(points_filtered, roads_filtered, '220')
```

```python
import osmnx as ox
import geopandas as gpd
import folium

# Город (замените на ваш, если нужно)
place = "Москва, Россия"

# 1. Загружаем граф дорог
G = ox.graph.graph_from_place(place, network_type='drive', simplify=True)
gdf_nodes, gdf_edges = ox.graph_to_gdfs(G)

# 2. Загружаем маршруты общественного транспорта и остановки
tags_routes = {'route': ['bus', 'tram', 'subway', 'train', 'trolleybus']}
gdf_routes = ox.features.features_from_place(place, tags=tags_routes)

tags_stops = {'highway': 'bus_stop', 'public_transport': 'stop_position'}
gdf_stops = ox.features.features_from_place(place, tags=tags_stops)

# 3. Создаём интерактивную карту с folium
center = gdf_edges.total_bounds[[1, 0]].values.tolist()  # [lat, lon]
m = folium.Map(location=center, zoom_start=12)

# Добавляем слои: дороги, маршруты, остановки
folium.GeoJson(
    gdf_edges,
    name="Дороги",
    style_function=lambda x: {'color': 'gray', 'weight': 1}
).add_to(m)

folium.GeoJson(
    gdf_routes,
    name="Маршруты",
    tooltip=folium.GeoJsonTooltip(fields=['name', 'route']),
    style_function=lambda x: {'color': 'blue', 'weight': 2}
).add_to(m)

folium.GeoJson(
    gdf_stops,
    name="Остановки",
    marker=folium.CircleMarker(radius=3, color='red', fill=True, fill_color='red')
).add_to(m)

# Добавляем управление слоями
folium.LayerControl().add_to(m)

# Выводим карту в Jupyter Notebook или сохраняем в переменной для просмотра
m  # Если не в Jupyter, раскомментируйте m.save('map.html') для сохранения в HTML
```
